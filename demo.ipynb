```python
# test 1: elongated bistar recovery algorithm
for n in range(5,12):
    ct = 0
    for comp in Compositions(n,length = 3):
        if not membership(comp):
            continue
        a,r,b = comp
        alpha = [a] + [1] * r + [b]
        X = caterpillar(alpha)
        f = qtchromatic(X,3)
        recovered = recover_elongated(f, verbose = False)
        ct += 1
        if comp != recovered:
            print(comp,recovered)
    print(f"Done with elongated bi-stars with {n} vertices and 3 interval vertices, total = {ct}.")
```

```python
# test 2: caterpillars with three invernal vertices recovery algorithm
for n in range(5,10):
    ct = 0
    for alpha in Compositions(n):
        if len(alpha) != 3:
            continue
        if not membership(alpha):
            continue
        X = caterpillar(alpha)
        f = qtchromatic(X,3)
        check_recover = recover_three_internal(f)
        ct += 1
        if not (check_recover == list(alpha) or check_recover[::-1] == list(alpha)):
            print(alpha, check_recover)
    print(f"Done with caterpillars with {n} vertices and 3 interval vertices, total = {ct}.")
```
